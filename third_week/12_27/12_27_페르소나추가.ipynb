```python
# ==========================================
# Persona Meta CSV 생성 전용 스크립트
# - 임베딩 없음
# - 계산 없음
# - CSV 템플릿만 생성
# ==========================================

import pandas as pd
from pathlib import Path

# ----------------------------
# 설정
# ----------------------------
OUTPUT_DIR = Path("./data")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_PATH = OUTPUT_DIR / "persona_meta_v2.csv"
PERSONA_COUNT = 8

# ----------------------------
# 컬럼 정의 (최대 확장)
# ----------------------------
COLUMNS = [
    # 식별
    "persona_id",
    "persona_name",

    # 핵심 (요구사항)
    "preference",            # 취향
    "shopping_pattern",      # 쇼핑 패턴
    "lifestyle",             # 라이프스타일

    # 피부/제품 적합 신호
    "skin_type",
    "skin_concern",
    "allergy_sensitivity",
    "texture_preference",
    "finish_preference",
    "scent_preference",

    # 사용 맥락/루틴
    "routine_step_count",
    "time_of_use",
    "seasonality",
    "environment_context",

    # 소비 성향/구매 행동
    "price_sensitivity",
    "brand_loyalty",
    "repurchase_tendency",
    "shopping_channel",
    "review_dependency",
    "bundle_preference",

    # 리스크/제약
    "ingredient_avoid_list",
    "ethical_preference",
    "treatment_status",

    # 메시지 개인화
    "message_tone_preference",
    "message_length_preference",
    "cta_style",
]

# ----------------------------
# CSV 생성
# ----------------------------
rows = []
for i in range(1, PERSONA_COUNT + 1):
    row = {col: "" for col in COLUMNS}
    row["persona_id"] = f"persona_{i}"
    row["persona_name"] = f"persona_{i}"
    rows.append(row)

df = pd.DataFrame(rows, columns=COLUMNS)
df.to_csv(OUTPUT_PATH, index=False, encoding="utf-8-sig")

print("=" * 50)
print("persona_meta_v2.csv 생성 완료")
print(f"경로: {OUTPUT_PATH}")
print(f"persona 수: {PERSONA_COUNT}")
print("→ 이 파일 채운 뒤, 임베딩은 별도 py에서 진행하세요.")
print("=" * 50)
```
