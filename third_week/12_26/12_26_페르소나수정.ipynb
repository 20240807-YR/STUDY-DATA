```python
import numpy as np
import pandas as pd

# 기존 페르소나 벡터 로드
base = np.load("../data_csv/persona_vectors.npy")  # (3, D)
assert base.shape[0] == 3

p1, p2, p3 = base

persona_vecs = {
    # 기존 3개 (유지)
    "persona_1": p1,
    "persona_2": p2,
    "persona_3": p3,

    # 확장 페르소나 (선형 결합)
    "persona_4": 0.7*p1 + 0.3*p2,        # 기능 + 감성
    "persona_5": 0.5*p2 + 0.5*p3,        # 감성 + 프리미엄
    "persona_6": 0.6*p1 + 0.4*p3,        # 기능 + 프리미엄
    "persona_7": (p1 + p2 + p3) / 3.0,   # 브랜드 이동형 (중립)
    "persona_8": p2 * 0.8 + np.random.normal(0, 0.01, p2.shape)  # 이탈/복귀 (약간 변형)
}

# 정규화 (중요)
def normalize(v):
    return v / np.linalg.norm(v)

persona_ids = list(persona_vecs.keys())
P = np.vstack([normalize(persona_vecs[k]) for k in persona_ids])

# 저장
np.save("../data_csv/persona_vectors.npy", P)
pd.DataFrame({"persona_id": persona_ids}) \
  .to_csv("../data_csv/persona_vectors.csv", index=False)
```
