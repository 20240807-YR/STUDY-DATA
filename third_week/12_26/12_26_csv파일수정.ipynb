```python
import pandas as pd

def load_crm_result(csv_path: str) -> pd.DataFrame:
    df = pd.read_csv(csv_path)

    REQUIRED_COLS = [
        "persona_id",
        "brand",
        "brand_tone_cluster",
        "avg_similarity",
        "product_cnt"
    ]

    # 컬럼 검증
    missing = set(REQUIRED_COLS) - set(df.columns)
    if missing:
        raise ValueError(f"CRM CSV missing columns: {missing}")

    return df[REQUIRED_COLS]
```
