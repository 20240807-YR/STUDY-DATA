```python
import pandas as pd
import re

CSV_PATH = "amore_basic_fields.csv"
COL_INGR = "전성분"

df = pd.read_csv(CSV_PATH)

assert COL_INGR in df.columns, f"'{COL_INGR}' 컬럼 없음. 컬럼명 다시 확인해라."

all_ingredients_raw = df[COL_INGR].dropna().astype(str).tolist()

ingredient_set = set()

for line in all_ingredients_raw:
   
    line = line.replace('"', '').replace("'", "")
    
 
    parts = line.split(",")
    
    for p in parts:
        ing = p.strip()
        
       
        if not ing:
            continue
        

        if re.fullmatch(r"[0-9.%()\s\-]+", ing):
            continue
        
        ingredient_set.add(ing)


ingredient_list = sorted(ingredient_set)

print(f"총 유니크 성분 수: {len(ingredient_list)}")
print("예시 30개:")
print(ingredient_list[:30])

out_df = pd.DataFrame({
    "ingredient_name": ingredient_list
})

out_df.to_csv("unique_ingredients.csv", index=False, encoding="utf-8-sig")

print("✅ unique_ingredients.csv 저장 완료")
```
