# 📊 브랜드 톤 분석 결과 리포트
이 노트북은 자동으로 생성된 분석 결과입니다.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 저장된 CSV 파일 로드
df = pd.read_csv('brand_analysis_result.csv')

# 데이터 상위 5개 확인
df.head()
```

```python
# 톤 분포 시각화
plt.figure(figsize=(10, 6))
sns.countplot(y='AI 분석 톤', data=df, order=df['AI 분석 톤'].value_counts().index, palette='viridis')
plt.title('전체 브랜드 AI 분석 톤 분포')
plt.show()
```

```python
# 브랜드별 지배적인 톤 확인
print(df.groupby('브랜드')['AI 분석 톤'].agg(lambda x: x.mode()[0]))

print("\n[톤별 평균 확신도]")
print(df.groupby('AI 분석 톤')['확신도'].mean().sort_values(ascending=False))
```
