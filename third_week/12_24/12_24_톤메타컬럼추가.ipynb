```python
import pandas as pd

# 기존 tone metadata 로드
df = pd.read_csv("../data_csv/tone_metadata.csv")

# 추가할 컬럼과 기본값 정의
new_columns = {
    "dominant_trait": "",
    "proof_level": "",
    "emotion_level": "",
    "cta_strength": "",
    "lexicon_group": "",
    "ban_group": "",
    "description": ""
}

# 컬럼 추가 (이미 있으면 건너뜀)
for col, default in new_columns.items():
    if col not in df.columns:
        df[col] = default

# 저장 (덮어쓰기 or 새 파일)
df.to_csv("tone_metadata_extended.csv", index=False)

print("✅ tone_metadata_extended.csv 생성 완료")
print(df.columns.tolist())
```
