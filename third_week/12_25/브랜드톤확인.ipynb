```python
import pandas as pd
import numpy as np

pkl_path = "../data_csv/brand_analysis_result.pkl"  # 필요하면 경로만 수정
df = pd.read_pickle(pkl_path)

# 1) 특정 브랜드만 뽑기 (컬럼명이 brand일 때)
brand_name = "라네즈"
b = df[df["brand"] == brand_name].copy()

# 2) tone_vector 생성 방식
# (A) 브랜드에 임베딩이 1개면 그대로
if len(b) == 1:
    tone_vector = np.array(b.iloc[0]["embedding_vector"], dtype=float)

# (B) 브랜드에 여러 row(파트/세그먼트)면 평균(centroid)으로 합치기
else:
    vecs = np.vstack(b["embedding_vector"].apply(lambda x: np.array(x, dtype=float)).to_list())
    tone_vector = vecs.mean(axis=0)

tone_vector  # np.ndarray (D,)
```
