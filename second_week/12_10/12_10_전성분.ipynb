```python
import pandas as pd

SRC_CSV = "amore_with_category.csv"
OUT_CSV = "amore_basic_fields.csv"

COL_NAME_PRODUCT = "상품명"
COL_NAME_BRAND   = "brand"
COL_NAME_INGR    = "전성분"

def main():
    df = pd.read_csv(SRC_CSV)

    # 필요한 컬럼만 추출 (없으면 빈 문자열로 채움)
    for col in [COL_NAME_PRODUCT, COL_NAME_BRAND, COL_NAME_INGR]:
        if col not in df.columns:
            df[col] = ""

    out_df = df[[COL_NAME_PRODUCT, COL_NAME_BRAND, COL_NAME_INGR]].copy()

    out_df.to_csv(OUT_CSV, index=False, encoding="utf-8-sig")
    print(f"[INFO] 생성 완료: {OUT_CSV}")

if __name__ == "__main__":
    main()
```

#### 전성분 임베딩 -> 성분 유사도가 곧 제품 유사도이며 같은 카테고리라도 전성분이 제품 성격을 완전히 결정함
