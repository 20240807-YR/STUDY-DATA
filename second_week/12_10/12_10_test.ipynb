#### 제품 포지셔닝 임베딩 벡터 (추후 고객 페르소나 및 전성분 임베딩 필요)

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

emb = np.load("amore_product_embeddings.npy")
meta = pd.read_csv("amore_product_meta.csv")

sim = cosine_similarity(emb, emb)

# 100번쨰 제품이랑 유사한 거
idx = 100
top5 = sim[idx].argsort()[::-1][1:6]

print(meta.iloc[idx])
print("\n유사 제품 추천:")

for i in top5:
    print(meta.iloc[i]["상품명"], meta.iloc[i]["category"], meta.iloc[i]["subcategory"])
```
