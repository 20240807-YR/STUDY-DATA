```python
import requests
from bs4 import BeautifulSoup
import csv

url = "https://www.amoremall.com/kr/ko/display/brand/detail/all?brandSn=31&frontNo=97&sort=Bestselling"
res = requests.get(url, headers={"User-Agent": "Mozilla/5.0"})
soup = BeautifulSoup(res.text, "html.parser")

cards = soup.select("div.productCard.shop-productCard.typeSmall")

rows = []

for c in cards[:30]:
    name = c.select_one(".shop-priceInfo .name")
    price = c.select_one(".shop-priceInfo .price")
    discount = c.select_one(".shop-priceInfo .discount")
    review = c.select_one(".shop-priceInfo .review")

    rows.append([
        name.get_text(strip=True) if name else "",
        price.get_text(strip=True) if price else "",
        discount.get_text(strip=True) if discount else "",
        review.get_text(strip=True) if review else "",
    ])

with open("laneige_bestselling.csv", "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    writer.writerow(["상품명", "가격", "할인율", "평점"])
    writer.writerows(rows)

print("CSV 저장 완료")
```
